3. Optimization

Question:

Assume there is a physical system defined by n variables and with and energy E that depends on these variables. Near a minimum of that energy, one can expand the energy into a Taylor series that then looks like

$E(x^i) = 1/2 A_{ij} x^i x^j + b_i x^i + c$

where higher order terms have been omitted, with a symmetric positive definite (spd) matrix A, a vector b, and a constant c.

3.a. Convert this task (looking for a minimum of the energy) into a linear system.

3.b. Write a Julia function that finds the location of the minimum, taking A, b, and c as input. NOTE: You do not need to implement an algorithm (such as the power method or Gaussian elimination), use the standard methods available in Julia.

3.c. You are afraid that your friends will try to fool you by calling your function with a matrix A that is not spd. Extend your function to check that the matrix A is indeed spd.

3.d. Test your function with a non-trivial physical system that has at least 10^6 unknowns. For example, you could assume the energy

$E(x^i) = \sum_i(x^i + x^{i-1} + \sin(\pi i / n))^2$

and define $x^0 = 0$. You do not need to use this energy, you can define your own system.

Answer:

3. (a):

$E(x^i)=1/2\  x^T\cdot A\cdot x + b^T\cdot x + c, \ x=(x_1,\dots, x_n), \ b=(b_1,\dots, b_n)$

3. (b):

$y=Dx-p, \ y^Ty + \lambda=1/2\  x^T\cdot A\cdot x + b^T\cdot x + c=x^T(D^TD)x-2(p^TD)x+p^Tp+\lambda=E(x)$, solve $D^TD=1/2\ A, -2p^T D=b$.
y=0 is the minimum for E(x), solve $Dx=p$, which is equivalent to $x=-1/2(A)^{-1}b$.

```julia
# 3.(b)
function minimum(A,b)
    return -1/2*b\A
end
```

```julia
# 3.(c) 
function check_spd(A)
    if A!=A'
        return false
    else
        D=eigen(A).values
        for i in D
            if i < 0
                return false
            end
        end
    end
    return true
end
```

```julia
# 3.(d) solve the minimum of E(x^i) = \sum_i(x^i + x^{i-1} + \sin(\pi i / n))^2, with n variables x^i. A is a n*n matrix, i ranges from 1 to n.

function gen_spd(n) #generate the spd matrix associated with E(x) with n variables
    A=zeros(n,n)
    for i in 1:n
        if i==1
            A[i,i]=1
            A[i,i+1]=1
        elseif i==n
            A[i,i]=1
            A[i,i-1]=1
        else
            A[i,i]=2
            A[i,i-1]=1
            A[i,i+1]=1
        end
    end
    return A
end
```

```julia
function gen_vec(n) # generate the vector b associated with E(x) with n variables
    b=zeros(n)
    for i in 1:n
        if i==1
            b[i]=2*(sin(pi*i/n)+sin(pi*(i+1)/n))
        elseif i==n
            b[i]=2*(sin(pi*i/n)+sin(pi*(i-1)/n))
        else
            b[i]=2*(sin(pi*i/n)+sin(pi*(i+1)/n)+sin(pi*(i-1)/n))
        end
    end
    return b
end
```

```julia
n = 10^6
A=gen_spd(n)
```

```julia
println(check_spd(A))
```

```julia
b=gen_vec(n);
```

```julia
x_min=minimum(A,b)
```
